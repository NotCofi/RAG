```python
import os
from dotenv import load_dotenv
from langchain_core.runnables.base import RunnableEach, RunnableSerializable

load_dotenv()

from rag.rag_engine import build_pipeline

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

retriever = build_pipeline("data/sample.pdf")

llm = ChatOpenAI(
    model="deepseek/deepseek-v4.1-flash",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
    temperature=0,
    max_retries=3
)



prompt_template = ChatPromptTemplate.from_template("""
You are a factual business analysis assistant.
Answer the query using ONLY the verified context text provided below.

CONTEXT:
{context}

QUERY:
{question}
""")

chain_inputs = {"context": retriever, "question": RunnablePassthrough()}

rag_chain: RunnableSerializable = (
    chain_inputs
    | prompt_template
    | llm
    | StrOutputParser()
)

result = rag_chain.invoke("What are the findings in this document?")
print(result)



```
